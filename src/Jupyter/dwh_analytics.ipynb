```python
import duckdb
import numpy as np
import pandas as pd
```

**Duckdbの接続(メモリ)**

```python
conn = duckdb.connect(':memory:')
```

**ダミーデータ生成**

```python
data = np.random.rand(100,100)
df = pd.DataFrame(data, columns=[f'col_{i}' for i in range(1, 101)])
df.head()
```

**DuckDBへの読み込み**

```python
conn.register("test_table", df)
```

**DuckDBから読み取り**

```python
result = conn.execute("SELECT col_1, col_2, col_3 FROM test_table WHERE col_1 > 0.5").fetchdf()
print(result)
```

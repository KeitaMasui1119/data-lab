**データの読み込み**

```python
cols = df_occto.columns
unpivot_cols = [col for col in cols if col not in ["発電所コード", "エリア", "発電所名", "ユニット名", "発電方式・燃種", "対象日", "更新日時"]]
print(unpivot_cols)
```

## データ加工

### 横持ち→縦持ち変換

```python
df_occto_melted = df_occto.unpivot(index=["発電所コード", "エリア", "発電所名", "ユニット名", "発電方式・燃種", "対象日", "更新日時"], on = unpivot_cols, variable_name="時間帯", value_name="供給力[kWh]")
print(df_occto_melted.shape)
df_occto_melted.head()
```

### 不要なデータの除外

```python
df_occto_filtered = df_occto_melted.filter(pl.col("時間帯") != "日量[kWh]")
print(df_occto_filtered.shape)
df_occto_filtered.head()
```

### 対象日時カラムの作成

```python
df_occto_dt = df_occto_filtered.with_columns(
    pl.when(pl.col("時間帯").str.contains("24:00"))
    .then(
        # 24:00の場合、対象日に1日加算して00:00に変換する
        (
            pl.col("対象日").str.to_date(format="%Y/%m/%d") + pl.duration(days=1)
        ).cast(pl.Utf8) + pl.lit(" 00:00")
    )
    .otherwise(
        # 24:00が含まれない場合: 元の文字列 (対象日 + 時間帯) を使用
        pl.col('対象日') + pl.lit(' ') + pl.col('時間帯').str.replace(r'\"?\[kWh\]\"?', '').str.strip_chars()
    ).alias("対象日時_str")
).with_columns(
    # 日時文字列をdatetime型に変換する
    pl.col("対象日時_str").str.to_datetime(format="%Y/%m/%d %H:%M").alias("対象日時")
).drop(["時刻","対象日時_str","対象日"])

print(df_occto_dt.shape)
df_occto_dt.head()
```
